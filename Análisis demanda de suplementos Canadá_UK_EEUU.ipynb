```python
# Importamos pandas y cargamos la tabla de datos.
import pandas as pd  
ruta= r"D:\Supplement_Sales_Weekly_Expanded.csv" 
df=pd.read_csv(ruta)
print(df.tail(40))
```

```python
# Mostramos los máximos valores de la tabla.

plataforma_mas_comun = df["Platform"].value_counts().idxmax()
print(f"La plataforma con mayores ventas es {plataforma_mas_comun}")
producto_mas_comun = df["Product Name"].value_counts().idxmax()
print(f"El producto más vendido es {producto_mas_comun}")
suplemento_mas_comun = df["Category"].value_counts().idxmax()
print(f"El suplemento más común es {suplemento_mas_comun}")
país_mas_ventas = df["Location"].value_counts().idxmax()
print(f"El país con más productos vendidos es {país_mas_ventas}")
mas_ventas=df["Units Sold"].value_counts().idxmax()
print(f"La mayor cifra de unidades vendidas es {mas_ventas}")
print(f"La plataforma con mayor volumen de ventas es {plataforma_mas_comun} con {mas_ventas} millones .")
```

```python
# Mostramos los valores minimos de la tabla.

plataforma_menos_comun = df["Platform"].value_counts().idxmin()
print(f"La plataforma con menores ventas es {plataforma_menos_comun}")
producto_menos_comun = df["Product Name"].value_counts().idxmin()
print(f"El producto menos vendido es {producto_menos_comun}")
suplemento_menos_comun = df["Category"].value_counts().idxmin()
print(f"El suplemento menos común es {suplemento_menos_comun}")
país_menos_ventas = df["Location"].value_counts().idxmin()
print(f"El país con menos productos vendidos es {país_menos_ventas}")
menos_ventas=df["Units Sold"].value_counts().idxmin()
print(f"La menor cifra de unidades vendidas es {menos_ventas} millones")
print(f"La plataforma con mayor volumen de ventas es {plataforma_menos_comun} con {menos_ventas} millones .")
```

```python
# Hacemos una lista con los 5 suplementos más vendidos y sus cifras de venta.
print("Los suplementos más vendidos son: ")

df.groupby('Category')['Units Sold'].sum().sort_values(ascending=False).head(7)


```

```python
# Ya sabemos que en Canadá lo que más se consume son minerales y vitaminas de procedencia natural pero, ahora vamos a ver qué suplementos son los que poseen una mayor demanda en EEUU.



# Filtrar por ubicación en EE. UU.
df_us = df[df["Location"] == "USA"]

# Obtener los 3 tipos de suplementos más vendidos
top_categories = df_us.groupby("Category")["Units Sold"].sum().sort_values(ascending=False).head(3)

# Obtener las 3 empresas más vendedoras
top_platforms = df_us.groupby("Platform")["Units Sold"].sum().sort_values(ascending=False).head(3)

print("Top 3 categorías:")
print(top_categories)

print("\nTop 3 plataformas:")
print(top_platforms)


```

El auge del mercado de suplementos alimenticios en Norteamérica
Norteamérica concentra aproximadamente el 33 % del mercado mundial de suplementos alimenticios, posicionándose como una de las regiones con mayor consumo. Dentro de este mercado, los productos más comercializados son los suplementos herbales y botánicos, seguidos por vitaminas y minerales, reflejando una tendencia hacia la prevención y el cuidado integral de la salud mediante compuestos naturales.

Tendencias de consumo en EE. UU. y Canadá
La salud inmunológica se ha consolidado como el principal motor de consumo en la región, impulsada especialmente por los efectos de la pandemia de COVID-19. Sin embargo, otras áreas de interés han emergido con fuerza, como la salud mental, la energía y la resistencia física, la salud digestiva y hepática, y el cuidado de la piel.

En Estados Unidos, los productos con beneficios asociados a la agudeza mental, la salud de las articulaciones y el tratamiento del insomnio presentan un crecimiento notable. En Canadá, se destacan las afirmaciones relativas a la salud cardiovascular, además de las ya mencionadas en el contexto estadounidense. Este panorama sugiere una transición hacia un consumo más personalizado, en el que los consumidores buscan respuestas específicas a necesidades particulares de salud.

Regulación canadiense de los suplementos alimenticios:
En Canadá, los suplementos están regulados bajo la categoría de Productos Naturales para la Salud (Natural Health Products, NHPs), supervisados por Health Canada. Esta categoría incluye vitaminas, minerales, extractos botánicos y otros compuestos naturales que no están clasificados como fármacos.

Tradicionalmente, estos productos se han regulado de forma independiente del sector farmacéutico. No obstante, la reciente aprobación del Proyecto de Ley C-47 en 2023 establece que los NHPs pasarán a estar bajo la jurisdicción de la Ley de Alimentos y Medicamentos, lo que implicará estándares más estrictos en cuanto a control de calidad, etiquetado y evidencia científica. Hasta que esta reforma se implemente por completo, los suplementos como el inositol seguirán estando sujetos a un marco regulatorio más flexible, similar al de la vitamina D o los complejos prenatales.

Esta flexibilidad normativa, sumada a una población cada vez más interesada en alternativas naturales, ha favorecido el crecimiento sostenido del consumo de suplementos en Canadá.

Legislación y consumo en Estados Unidos:
En contraste, en Estados Unidos la regulación del mercado de suplementos corre a cargo de la Food and Drug Administration (FDA). Sin embargo, su enfoque se limita principalmente al etiquetado y la inocuidad de los productos, dejando la responsabilidad de comprobar la eficacia y seguridad en manos de los fabricantes.

Este marco regulatorio relativamente permisivo ha facilitado la expansión del mercado, permitiendo la introducción rápida de nuevos productos sin necesidad de estudios clínicos rigurosos previos. Este entorno ha favorecido un comportamiento de consumo en el que los suplementos forman parte de la vida cotidiana de millones de personas.

Además, el estilo de vida moderno, caracterizado por una alimentación procesada, estrés crónico y falta de actividad física, ha impulsado la percepción de los suplementos como una solución rápida para cubrir carencias nutricionales o tratar síntomas comunes.


Tanto en Canadá como en EE. UU., el consumo de suplementos se encuentra en una fase de crecimiento sostenido, influido por factores culturales, comerciales y regulatorios. Si bien las diferencias en la legislación entre ambos países son notables, en ambos casos se observa una tendencia clara hacia la búsqueda de soluciones naturales para mejorar la calidad de vida.

No obstante, el bajo nivel de exigencia en cuanto a la evidencia científica de muchos suplementos plantea desafíos importantes en términos de seguridad, eficacia y ética de comercialización, especialmente a medida que se difuminan las fronteras entre productos nutricionales y terapéuticos.

```python
# A continuación vamos a hacer una gráfica que muestre el creciente consumo de Vitamina, Minerales y Proteinas en los uúltimos 5 años.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Cambiamos la fecha a date time
df['Date'] = pd.to_datetime(df['Date'])

# Filtrar por categoría Vitamin y Mineral, y por países Canada y USA
filtered_df = df[
    df['Category'].isin(['Vitamin', 'Mineral']) &
    df['Location'].isin(['Canada', 'USA'])
]

# Agrupar por año, categoría y país, y sumar unidades vendidas
filtered_df['Year'] = filtered_df['Date'].dt.year
summary = filtered_df.groupby(['Year', 'Category', 'Location'])['Units Sold'].sum().reset_index()

# Crear gráfico de tabla
plt.figure(figsize=(12, 6))
sns.lineplot(data=summary, x='Year', y='Units Sold', hue='Category', style='Location', markers=True, dashes=False)
plt.title('Demanda de Vitaminas y Minerales en Canadá y EE.UU. (Últimos 5 años)')
plt.ylabel('Unidades Vendidas')
plt.xlabel('Año')
plt.grid(True)
plt.tight_layout()
plt.show()

```

Podemos observar cómo entre 2023 y 2024 se alcanza un pico en las demandas de ambos suplementos debido a la cercanía de la pandemia y la creciente preocupación de la sociedad por la salud y el bienestar. Además, las presiones de las campañas de marketing y las regulaciones mencionadas en el comentario anterior favorecieron el consumo de estos suplementos. Sin embargo, en el siguiente año, los factores económicos como la crecietne inflación y la crisis ocasionada por la pandemia o el exceso de oferta producido en este sector del mercado, dieron lugar a un gran decrecimiento en este último periodo. Otros factores que han podido contribuir a este claro descenso han podido ser la desconfianza en este sector o el desccenso en las investigaciones o simplemetne la pérdida de interés por parte del consumidor.

En resumen, los factores sociopolíticos, la pandemia del coronavirus y las consecuencias económicas así como las tendencias en las investigaciones han sido los principales factores explicativos de la variación de la demanda del sector de la suplementación alimenticia en Norteamérica, además, los pronósticos de cara al futuro no son alcistas ya que se podría producir un proceso de estabilización o de decrecimiento en suplementos como la proteína, la vitamina C o los minerales.
